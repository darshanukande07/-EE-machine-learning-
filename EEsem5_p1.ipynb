```python

# For categorical columns
cat_cols = df.select_dtypes(include='object').columns
df[cat_cols] = df[cat_cols].fillna(df[cat_cols].mode().iloc[0])
```

```python
# For numerical columns
num_cols = df.select_dtypes(include=np.number).columns
df[num_cols] = df[num_cols].fillna(df[num_cols].mean())
```

```python
# Convert categorical data into numerical form
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

for col in cat_cols:
    df[col] = encoder.fit_transform(df[col])
```

```python
# Feature scaling
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

df[num_cols] = scaler.fit_transform(df[num_cols])
```

```python
#Check outliers using IQR method
for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    df = df[(df[col] >= lower_limit) & (df[col] <= upper_limit)]
```

```python
print("Minimum :", df['Total Amount'].min())
print("Q1 :", df['Total Amount'].quantile(0.25))
print("Median :", df['Total Amount'].median())
print("Q3 :", df['Total Amount'].quantile(0.75))
print("Maximum :", df['Total Amount'].max())
```

```python
print("Mean :", df['Total Amount'].mean())
print("Median :", df['Total Amount'].median())
print("Mode :")
print(df['Total Amount'].mode())
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

df[['Age','Price per Unit','Quantity','Total Amount']] = scaler.fit_transform(
    df[['Age','Price per Unit','Quantity','Total Amount']]
)

print(df.head())
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df[['Age','Price per Unit','Quantity','Total Amount']] = scaler.fit_transform(
    df[['Age','Price per Unit','Quantity','Total Amount']]
)

print(df.head())
```

```python
bins = [0,20,40,60,100]
labels = ['Teen','Young','Adult','Senior']

df['Age Group'] = pd.cut(df['Age'],
                         bins=bins,
                         labels=labels)

print(df[['Age','Age Group']].head())
```

```python
df['Amount_per_Item'] = df['Total Amount'] / df['Quantity']

print(df[['Amount_per_Item']].head())
```

```python
df['Log_Total_Amount'] = np.log1p(df['Total Amount'])

print(df[['Total Amount','Log_Total_Amount']].head())
```
